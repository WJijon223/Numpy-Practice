## Exercise 1 (Employee Pay)

```python
import numpy as np

# Number of hours worked by 8 employees (and hourly rate)
hours = np.array([35, 42, 38, 50, 29, 45, 40, 33])
hourly_rate = 22.50

pay = np.array(hours * hourly_rate)
print(f"Total pay for each employee: {np.round(pay, decimals=2)}")

print(f"Employees' pay amounts that were greater than 900: {pay[pay > 900]}")

employee_nums = np.arange(1, 9)
print(f"Employee IDs: {employee_nums}")

print(f"IDs of employees who worked 40 or more hours: {employee_nums[hours[employee_nums - 1] >= 40]}")

new_arr = np.where(hours < 40, hours, 40)
print(new_arr)
```

## Exercise 2 (Exam Gradebook)

```python
grades = np.array([
    [85, 92, 78, 90],
    [76, 88, 95, 82],
    [91, 79, 87, 94],
    [68, 74, 80, 77],
    [95, 89, 92, 96]
])

print(grades.dtype)
print(grades.shape)

print(f"All exam grades for student 4: {grades[3]}")

print(f"Exam 3 scores for every student: {grades[:, 2]}")

print(f"Grades for first 3 students on Exams 2 through 4: {grades[:3, 1: 4]}")
```
